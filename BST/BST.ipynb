```python
class User:
    def __init__(self, username, name, email):
        self.username = username
        self.name = name
        self.email = email

    def __repr__(self):
        return f"User(username='{self.username}', name='{self.name}', email='{self.email}')"

class UserDatabase:  # ✅ Class names should be PascalCase
    def __init__(self):
        self.users = []

    def insert(self, user):
        i = 0
        while i < len(self.users):
            if self.users[i].username > user.username:  # ✅ Compare usernames
                break
            i += 1
        self.users.insert(i, user)  # ✅ Insert at the correct index

    def find(self, username):
        for user in self.users:
            if user.username == username:
                return user  # ✅ Return the user object, not the username
        return None  # ✅ Return None if not found

    def update(self, user):
        target = self.find(user.username)
        if target:  # ✅ Check if the user exists
            target.name = user.name
            target.email = user.email

    def list_all(self):
        return self.users
```

```python
def tree_to_tuple(node):
    print(node.key)
    # case 1 : if node is empty 
    if node is None:
        return None

    # case 2: if leaf node doesn't have child no right and no left child
    if node.right is None and node.left is None:
        return node.key
    else:# that means have a child
        return(
            tree_to_tuple(node.left),node.key,tree_to_tuple(node.right))
        
```
