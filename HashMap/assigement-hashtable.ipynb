## Hashing Function
A hashing function is used to convert strings and other non-numeric data types into numbers, which can then be used as list indices. For instance, if a hashing function converts the string "Aakash" into the number 4, then the key-value pair 'Aakash': '7878787878' will be stored at the position 4 within the data list.

1.Iterate over the string, character by character
2.Convert each character to a number using the ord function.
3.Add the numbers for each character to obtain the hash for the entire string
4.Take the remainder of the result with the size of the data list

```python
#code 
def get_index(data_list, string):
    count=0
    for a in string:
        no=ord(a) ## convert string to number

        count+=no

    list_index= count%len(data_list) ## fit into len of datalist , vaild data list
    return list_index
    
```

## Now we ready to build basic implementation of hashtable

```python
class basichashtable:
    def __init__(self,mx=Mx_hash_table_size):
        self.data_list=[None]*mx

    def get_index(self,string):
        count=0
        for char in string:
            count+=ord(char)
        list_index=count%len(self.data_list)
        return list_index


    def insert(self,key,value):
        idx=self.get_index(key)

        self.data_list[idx] = key,value

    # update 
    def update(self,key,value):
        idx=self.get_index(key)

        self.data_list[idx]=key,value

    def find(self,key):
        idx=self.get_index(key)

        kv=self.data_list[idx]

        if kv is None:
            return None
        else:
            key,value=kv
            return value
    ## list all 
    def list_all(self):
        kv =[i for i in self.data_list if i is not None]
        return kv


    ## deletion 
    def delete(self,key):
        idx=self.get_index(key)# see the key by getting index

        kv = self.data_list[idx] # access the key-value from index

        if kv is None:
            return None
        else:
            new_key,new_value=kv
            if new_key == key:
                self.data_list[idx] =None
                
                return f"nigga your {new_key} and {new_value} was deleted"

            else:
                return None
            
           
        

        
```

### Hashing we have a promblem like in this we are just doing conversion string to number but what if the two different string have same character just order is different

#### For Example string 1 : "slient" and string 2 : "listen" have same character so hashing give same number for those and while assign the a index those two goes in the same index which is overlap this is the main issue instead allocation it update at same index

for encounter this problem we have to think a solution like we just move one index to allocation 

To handle collisions we'll use a technique called linear probing. Here's how it works:

1. While inserting a new key-value pair if the target index for a key is occupied by another key, then we try the next index, followed by the next and so on till we the closest empty location.
2. While finding a key-value pair, we apply the same strategy, but instead of searching for an empty location, we look for a location which contains a key-value pair with the matching key.
3. While updating a key-value pair, we apply the same strategy, but instead of searching for an empty location, we look for a location which contains a key-value pair with the matching key, and update its value.

```python
def get_valid_index(data_list,key):
    idx=get_index(data_list,key) # get the index 
    start_idx=idx

    while True:
        # get the key value pair for stored at idx
        kv=data_list[idx] # now we see data_list assign index store kv "key value"

        # if index have none value means nothing there just reutrn 
        if kv is None:
            return idx

        # if the stored key matched to given key, return the index
        k,v = kv

        # if key found in data list, key is already exist
        if k == key:
            return idx # found the key
        idx+=1

        if idx == len(data_list): # wrap around the list 0
            idx=0

        if idx==start_idx:
            raise Exception("data list is full")

def get_index(data_list,String):
    count=0

    for char in String:
        count+=ord(char)

    index=count%len(data_list)
    return index
        
            

        
```

```python
# Create an empty hash table
data_list = [None] * 1024

# New key 'listen' should return expected index
get_valid_index(data_list, 'listen') == 655
```

```python
# Insert a key-value pair for the key 'listen'
data_list[get_index(data_list, 'listen')] = ('listen', 99)

# Colliding key 'silent' should return next index
get_valid_index(data_list, 'silent') == 656
```
