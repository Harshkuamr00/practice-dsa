 ## lets do some exercise with hashmap 

### assigement to implement a hashtable class support the following operations:
1. insert
2. find
3. update
4. list


```python


class hashtable:
    def __init__(self):
        self.table={}
        
    def insert(self,key,value):
        self.table[key]=value

    def find(self,key):
        return self.table[key]

    def update(self,key,value):
        if key in self.table:
            self.table[key]=value
        else:
            print(f" the {key} is not found in hashmap")

    def delete(self,key):
        if key in self.table:
            del self.table[key]
        else:
            print(f" the {key} is not found in table")

    def list_all(self):
        return self.table.items()
```

```python
# Create a hash table
ht = hashtable()

# Insert data
ht.insert("name", "Alice")
ht.insert("age", 30)
ht.insert("city", "New York")

```

```python
ht.find("age")
```

```python
ht.update("city","pander")
```

```python
ht.delete("city")
```

```python
ht.list_all()
```
