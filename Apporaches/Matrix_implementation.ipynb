#### Question 1 : Dimensions (shape) of a Matrix

```python
def get_demensition(mat):
    if not mat:
        return (0,0)
    row=len(mat)
    col=len(mat[0]) if row > 0 else 0
    return (row,col)
```

```python
mat=Matrix
print(get_demensition(mat))
```

#### Question 2 : Get the size of matrix

```python
def size(mat):
    row,col= get_demensition(mat)
    return row * col
```

```python
print(size(mat))
```
