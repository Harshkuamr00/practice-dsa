### pointer has type it depends where to start the lef and right all tpyes menstion in below:

1. Converging Pointers
-> start with ending of the list or array left end and right end, compare with specific condition if met then stop and stop until l1 and R1 cross each other.

3. Parallel Pointers

-> in this apporach we move first pointer independently until it finds an element that meet a certain condition after this we start traversing with the secound pointer to addition information related to first pointer found.
this teqnice used when we have to process elements in stages

### Question 1: Two sum(sorted array)
Given a sorted array nums = [2, 7, 11, 15] and target = 9, return the indices of the two numbers that add up to target.

##### test Cases

```python
tests=[]
#test case 1
tests.append({
    "input":{
        "nums":[2,7,11,15],
        "target":9,
    },
    "output":[0,1]
})

# test case 2
tests.append({
    "input":{
        "nums":[2,3,4],
        "target":6,
    },
    "output":[0,2]
})
# test case 3
tests.append({
    "input":{
        "nums":[3,3],
        "target":6,
    },
    "output":[0,1]
})

# test case 4
tests.append({
    "input":{
        "nums":[3, 24, 50, 79, 88, 150, 345],
        "target":200,
    },
    "output":[2,5]
})

# test case 5
tests.append({
    "input":{
        "nums":[],
        "target":200,
    },
    "output":[]
})

# test case 6
tests.append({
    "input":{
        "nums":[3,3,3,3,3],
        "target":6,
    },
    "output":[0,1]
})

#test case 7
tests.append({
    "input":{
        "nums":[-1,0],
        "target":-1,
    },
    "output":[0,1]
})

    
```

```python
## code using conversing pointer
class solution:
    def __init__(self,nums,target):
        self.nums=nums
        self.target=target 
    ## two sum sorted array  
    def sol(self):
        if not self.nums:
            return []
        l,r=0,len(self.nums)-1

        ## taking loop
        while l<r:
            if self.nums[l]+self.nums[r]>self.target:
                r-=1
            elif self.nums[l]+self.nums[r]<self.target:
                l+=1
            else:
                result=[l,r]
                r-=1
                
        return result if result else []
        
                
                
                
```

```python
l1=solution([2,7,11,15], 9)
```

```python
print(l1.sol())
```

```python
for test in tests:
    current_test = solution(**test["input"])
    result = current_test.sol()
    print(f"Result: {result} | Expected: {test['output']} | Passed: {result == test['output']}")
```

### Question 2 : Remove Duplicates (in-place)
Brief: Modify the array in-place to remove duplicates and return the new length.

```python
tests=[]

tests.append({
    "input":{
        "nums":[1,1,2],
    },
    "Output":2
})

# test case 2
tests.append({
    "input":{
        "nums":[],
    },
    "Output":[]
})

# test case 3
tests.append({
    "input":{
        "nums":[-1,0,1,1],
    },
    "Output":3
})

# test case 4
tests.append({
    "input":{
        "nums":[1,2,3],
    },
    "Output":3
})
```

```python
## code 
class solution1:
    def __init__(self,nums):
        self.nums=nums

    # solution method
    def remove_dupilcates(self):
        l,r=0,len(self.nums)-1
        if not self.nums:
            return []

        while l<r:
            if self.nums[l]==self.nums[r]:
                self.nums.pop(l)
                l+=1
            elif self.nums[l]<self.nums[r]:
                r-=1
                
            else:
                r-=1
        return len(self.nums)      
            
```

### Question 3 : Return the squares of the array in sorted order.

Test case

```python
tests=[]
# test case 1
tests.append({
    "input":{
        "nums":[-4,-1,0,3,10],
    },
    "output":[0,1,9,16,100]
})

# test case 2
tests.append({
    "input":{
        "nums":[],
    },
    "output":[]
})

# test case 3
tests.append({
    "input":{
        "nums":[0,0,1,1,1,1],
    },
    "output":[0,0,1,1,1,1]
})

# test case 4
tests.append({
    "input":{
        "nums":[0,1,2,3,4],
    },
    "output":[0,1,4,9,16]
})
```

```python
## code 
class Question3:
    def __init__(self,nums):
        self.nums=nums

    def square(self):
        nums=self.nums
        if not nums:
            return []
        
        nums=[x**2 for x in nums]
        n=len(nums)
        ## make another list as same as nums
        result=n*[0]
        l,r=0,n-1
        index=n-1

        while l <= r:
            if nums[l] > nums[r] :
                result[index]=nums[l]
                l+=1
            else:
                result[index]=nums[r]
                r-=1
            index-=1
        return result
            
```

```python
l3=Question3([-4,-1,0,3,10])
```

```python
l3.square()
```

#### Question 2: Detect cycle in a linked list

Given a linked list with a cycle: 1 -> 2 -> 3 -> 4 -> 2 (cycle at node 2), detect if a cycle exists.

```python
class node:
    def __init__(self,num):
        self.data=num
        self.next=None


class Linkedlist:
    def __init__(self):
        self.head=None


    ## add method 
    def add(self,num):
        if self.head is None:
            self.head=node(num)

        else:
            current_node=self.head
            while current_node.next !=None:
                current_node=current_node.next
            current_node.next=node(num)

    ## print method
    def show(self):
        current=self.head
        while current !=None:
            print(current.data)
            current=current.next
    ## find middle element
    def middle(self):
        if self.head is None:
            return None

        slow=self.head
        fast=self.head

        while fast is not None and fast.next is not None:
            slow=slow.next
            fast=fast.next.next
        return slow.data

    ## find cycle 
    def cycle(self):
        if self.head is None:
            return None

        slow=self.head
        fast=self.head

        while fast is not None and fast.next is not None:
            slow=slow.next
            fast=fast.next.next

            if slow is fast:
                return True
        return False
        
        
            
    
```
