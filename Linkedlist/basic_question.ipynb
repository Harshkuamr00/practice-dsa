## practices question in Linkedlist

question 1: insert at the begining, we write a method a number insert at start of the linked list

```python
## code
class node():
    def __init__(self,num):
        self.data=num
        self.next=None

class linkedlist():
    def __init__(self):
        self.head=None

    def add(self,num):
        if self.head is None:
            self.head=node(num)
        else:
            current_node=self.head
            while current_node.next is not None:
                current_node=current_node.next
            current_node.next=node(num)

    # initize the method to add any value at the begning
    def add_first(self,num):
        new_node=node(num)
        new_node.next=self.head
        self.head=new_node

    ## initize the method to add any value to at the end of the list
    def add_end(self,num):
        new_node=node(num)
        if self.head==None:
            self.head=new_node
        else:
            current=self.head
            while current.next != None:
                current=current.next
            current.next=new_node

    def count(self):
        count=0
        current=self.head
        while current is not None:
            current=current.next
            count+=1
        return count

    def show(self):
        current=self.head
        while current is not None:
            print(current.data)
            current=current.next

    ## search a value in the linked list

    def find(self,target):
        position=0
        current = self.head
        while current is not None:
            if current.data==target:
                print(position)
            current=current.next
            position+=1
            
        
```

```python
list=linkedlist()
list.add(1)
list.add(2)
list.add(3)
list.add(4)

list.add_first(5)
list.add_end(7)
list.show()
```
