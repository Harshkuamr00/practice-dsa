## Intermidiate question 

doing some other question that have intermidate level of coding like deleting a node by value, insert after a given node, reverse linked list etc

```python
## code 
class node():
    def __init__(self,num):
        self.data=num
        self.next=None


class linkedlist():
    def __init__(self):
        self.head=None

    ### add method to add value
    def add(self,num):
        if self.head is None:
            self.head=node(num)
        else:
            current=self.head
            while current.next is not None:
                current=current.next
            current.next=node(num)

    ## print all value in linked list
    def show(self):
        current = self.head
        while current is not None:
            print(current.data)
            current=current.next

    ## delete a node by value
    def delete_node(self,num):
        number=0
        if self.head is None:
            return None

        if self.head.data==num:
            self.head=self.head.next
            return

        current=self.head
        while current.next is not None:
            if current.next.data==num:
                current.next=current.next.next
                return 
            current=current.next
        
    ## insert after a given node
    def inset_after(self,target,num):
        new_node=node(num)

        ## condition empty
        if self.head is None:
            return

        current = self.head
        while current is not None:
            if current.data == target:
                new_node.next=current.next
                current.next=new_node
            current=current.next

                
                
        
```

```python
list = linkedlist()
list.add(10)
list.add(20)
list.add(30)
list.add(40)
list.add(50)
list.add(60)
```

```python
list.show()
```

```python
list.delete_node(50)
```
