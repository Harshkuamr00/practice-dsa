#### Question 1 : Dimensions (shape) of a Matrix

```python
def get_demensition(mat):
    if not mat:
        return (0,0)
    row=len(mat)
    col=len(mat[0]) if row > 0 else 0
    return (row,col)
```

```python
mat=Matrix
print(get_demensition(mat))
```

#### Question 2 : Get the size of matrix

```python
def size(mat):
    row,col= get_demensition(mat)
    return row * col
```

### Traversal 
1. Row-wise Traversal : Left to right, top to bottom
2. column-wise Traversal : Top to bottom, left to right

```python
## row wise implementation
def traversal_row_wise(matrix):
    for i in matrix:
        for j in i:
            print(j, end=" ")
        print()
```

```python
print(traversal_row_wise(Matrix))
```

```python
## col wise implementation
def traversal_col_wise(matrix):
    if not matrix:
        return
    cols=len(matrix[0])
    for col in range(cols):
        for row in range(len(matrix)):
            print(matrix[row][col], end=" ")
        print()
```

```python
traversal_col_wise(Matrix)
```

#### Adding matrixes 

```python
## code 
def add_matrix(matrix1, matrix2):
    if not matrix1 or not matrix2:
        return []

    row1,col1=len(matrix1),len(matrix1[0])
    row2,col2=len(matrix2),len(matrix2[0])

    if row1 != row2 or col1!= col2:
        return []

    result =[]
    for i in range(row1):
        new_row=[]
        for j in range(col1):
            new_row.append(matrix1[i][j] + matrix2[i][j])
        result.append(new_row)
    return result

    

    
```

```python
matrix1 = [[1, 2], [3, 4]]
matrix2 = [[5, 6], [7, 8]]
print(add_matrix(matrix1, matrix2))
```

#### Substraction of two matrix

```python
## code 
def subs_mat(mat1,mat2):
    if not mat1 or not mat2:
        return []
    row1,col1=len(mat1),len(mat1[0])
    row2,col2=len(mat2),len(mat2[0])
    result=[]
    for i in range(row1):
        new_row=[]
        for j in range(col1):
            new_row.append(mat1[i][j] - mat2[i][j])
        result.append(new_row)
    return result
```

```python
matrix1 = [[1, 2], [3, 4]]
matrix2 = [[5, 6], [7, 8]]

print(subs_mat(matrix1, matrix2))
```

#### Scaler Multiplication


```python
## code 
def scaler_multi(matrix,k):
    if not matrix:
        return []

    result=[]
    for row in matrix:
        new_row=[i * k for i in row]
        result.append(new_row)
    return result
```

```python
print(scaler_multi(matrix1,2))
```

#### Matrix multiplication 

A = [
  [1, 2],
  [3, 4]
]
B = [
  [5, 6],
  [7, 8]
]
C = A * B
C[0][0] = 1*5 + 2*7 = 19
C[0][1] = 1*6 + 2*8 = 22
C[1][0] = 3*5 + 4*7 = 43
C[1][1] = 3*6 + 4*8 = 50
Result:
[
  [19, 22],
  [43, 50]
]



```python
## code 
def multiplication(mat1,mat2):
    if not mat1 or not mat2:
        return []

    row1,col1=len(mat1),len(mat1[0])
    row2,col2=len(mat2),len(mat2[0])

    if col2 != row2:
        return []

    result=[[0 for _ in range(col2)] for _ in range(row1)]
    for i in range(row1):
        for j in range(col2):
            for k in range(col1):
                result[i][j]+=mat1[i][j] * mat2[k][j]
    return result
```

```python
matrix1 = [[1, 2], [3, 4]]
matrix2 = [[5, 6], [7, 8]]
print(multiplication(matrix1, matrix2))
```
