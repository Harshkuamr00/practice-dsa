### VP network of user friendship

##### Define Users 

```python
users = [
{ "id": 0, "name": "Hero" },
{ "id": 1, "name": "Dunn" },
{ "id": 2, "name": "Sue" },
{ "id": 3, "name": "Chi" },
{ "id": 4, "name": "Thor" },
{ "id": 5, "name": "Clive" },
{ "id": 6, "name": "Hicks" },
{ "id": 7, "name": "Devin" },
{ "id": 8, "name": "Kate" },
{ "id": 9, "name": "Klein" }
]
```

##### Frineds pairs between them followed by tuple implementation 

```python
friendship_pairs = [(0, 1), (0, 2), (1, 2), (1, 3), (2, 3), (3, 4),
(4, 5), (5, 6), (5, 7), (6, 8), (7, 8), (8, 9)]
```

for example the tuple (0,1) means hero and dunn are friends

## lets print the name of users with ids that have relations with someonem

```python
friendships={user["id"] : []  for user in users}
```

```python
friendships_by_name={user['name'] : [] for user in users}
```

```python
id_to_name={user['id'] : user['name'] for user in users}
```

```python
friendships
```

```python
for i,j in friendship_pairs:
    friendships[i].append(j)
    friendships[j].append(i)
    friendships_by_name[id_to_name[i]].append(id_to_name[j])  # Use names instead of IDs
    friendships_by_name[id_to_name[j]].append(id_to_name[i])
```

```python
friendships_by_name
```

```python
friendships
```

### lets find the total number of connections by summing up the length of all the friends 

```python
def number_of_friends(user):
    user_id = user["id"]
    friend_ids = friendships[user_id]
    return len(friend_ids)
    
print(sum(number_of_friends(user)
for user in users))
```

### lest find indiviual number of friends

```python
indiviual_connection={user['name']  : number_of_friends(user) for user in users}
```

```python
indiviual_connection
```

```python
total_connection=sum(number_of_friends(user)
for user in users)
total_users=len(users)
```

#### Average users connection

```python
average_connection=total_connection/total_users
```

```python
average_connection
```

### lets divide into bucket 

```python
def tenure_bucket(t):
    if t < 2:
        return "Less than two"
    elif t <= 5:
        return "Between two and five"
    else:
        return "More than five"



buckets = {}
for salary, tenure in salaries_and_tenures:
    b = tenure_bucket(tenure)
    buckets.setdefault(b, []).append(salary)

avg_salary_by_bucket = {b: sum(sals)/len(sals) for b, sals in buckets.items()}
print(avg_salary_by_bucket)
```
