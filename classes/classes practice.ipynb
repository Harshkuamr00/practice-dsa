create a class of bank account

```python
## code
class BankAccount:
    def __init__(self,account_holder,balance=0):
        self.account_holder=account_holder
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount

    def withdraw(self,amount):
        if amount> self.balance:
            return 'Insuficient Balance'
        else:
            self.balance=self.balance-amount

    ## check balance 
    def check_balance(self):
        return self.balance

        
```

### Store Books in a Library:
create a class with a list of book object and add method like add_book,find_book_by_name,list_book

```python
class Book:
    count=0
    def __init__(self,book_name,title):
        self.book_name=book_name
        self.title=title
        Book.count+=1


class Library:
    def __init__(self):
        self.books=[]

      ## add books to library
    def add_book(self,nbook_name):
        i=0
        while i<len(self.books): ## or Book.count
            if self.books[i].book_name>nbook_name.book_name:
                break
            i+=1
        self.books.insert(i,nbook_name)

    ## find the book by title
    def find(self,bname):
        for book in self.books:
            if book.book_name==bname:
                return book
        return None

    def list_books(self):
        return self.books
        

  
        
        
```

```python
# Create a library
library = Library()

# Predefined test data
test_books = [
    Book(book_name="The Hobbit", title="An Unexpected Journey"),
    Book(book_name="1984", title="A Dystopian Future"),
    Book(book_name="The Lord of the Rings", title="The Fellowship of the Ring"),
    Book(book_name="Pride and Prejudice", title="A Classic Novel"),
    Book(book_name="To Kill a Mockingbird", title="A Story of Injustice")
]

# Insert all test books into the library
for book in test_books:
    library.add_book(book)

# Test the library methods
print("All books in the library:")
for book in library.list_books():
    print(f"Book Name: {book.book_name}, Title: {book.title}")

# Test find method
found_book = library.find("1984")
if found_book:
    print(f"\nFound book: Book Name: {found_book.book_name}, Title: {found_book.title}")
else:
    print("\nBook not found.")
```
