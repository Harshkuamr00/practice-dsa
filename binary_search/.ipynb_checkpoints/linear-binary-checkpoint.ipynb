 Question : Alice has some cards with numbers written on them. She arranges the cards in decreasing order, and lays them out face down in a sequence on a table. She challenges Bob to pick out the card containing a given number by turning over as few cards as possible. Write a function to help Bob locate the card.

input
1. cards: A list of numbers sorted in decreasing order. E.g. [13, 11, 10, 7, 4, 3, 1, 0]
2. query: A number, whose position in the array is to be determined. E.g. 7


Output
3.position: position of quesy in the cards eg 3 

All possible way or test cases
1. the nuber or query that we looking for have in someware in middle
2. query is the 1st element in cards
3. query is the last element in cards
4. the list contains only one card
5. list does not contains query
6. list is empty
7. the list cards contains repeating number
8. list of cards have more than one query

Problem breakdown
step: we have serveral cards in list query the and position of list 
1. we take position variable initialze with 0
2. while loop for iteration
3. check condition if cards[position]==query ; return position
4. iterate next cards so we increase position with plus 1
5. check condition if position == len(card) we exit the loop and cards doesn't contain query; return -1


```python
# code 
def find(cards,query):
    # postion variable
    position=0

    # iteration
    while True:

        # check condition for query 
        if cards[position]==query:
            return position

        # if can't we increase position by 1
        position+=1

        # checking end of the list of cards and can't find query we exit the loop and return -1
        if position==len(cards):
            return -1
```

## Complexity and Big O Notation


Complextity of an algo is a measure of the amt of time and or sapce required by an algoritm for an input of given size.

```python
 def find1(cards,query):
     lower=0
     high=len(cards)-1

     while lower<=high:
         mid=(lower+high)//2
         mid_no=cards[mid]
         
         if mid_no==query:
             return mid
         elif mid_no < query:
             high=mid-1
         elif mid_no>query:
             lower=mid +1
     return -1
     
```

```python
##debug
def test_location(cards,query, mid):
    if cards[mid]==query:
        if mid-1>=0 and cards[mid-1] ==query:
            return 'left'
        else:
            return 'found'
    elif mid<query:
        return  'left'
    else:
        return 'right'

def locate_card(cards,query):
    low,high=0,len(cards) -1

    while low <= high:
        mid=(low+high)//2 
        result=test_location(cards,query,mid)

        if result == 'found':
            return mid
        elif result == 'left':
            high = mid-1
        elif result =='high':
            low=mid+1
    return -1
    
        
```

```python
for test in tests:
    evaluate_test_case(locate_card,test)
```
