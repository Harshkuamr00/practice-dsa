## Given question 
Find the count of rotation of sorted list an unkown number of times, minimum number of times its rotation
eg. [1,2,3,4,5,6,7] -> [5,6,7,1,2,3,4] 
number of rotation will be : 3


### define the test cases:
1. length of list is 10 rotated by 3 times
2. length of list is 8 rotated by 8 times
3. list was empty
4. list containing once element
5. list doesn't rotated
6. list rotated n-1 times where the size of list is n
7. list rotated n times ( back to initial stage)

```python
tests=[]
tests.append({
    'input':{
        'nums':[8,10,11,1,2,3,4,5,6,7]
    },
    'output':3
})
```

```python
tests.append({
    'input':{
        'nums':[11,12,13,14,15,1,2,3]
    },
    'output':5
})
```

```python
tests.append({
    'input':{
        'nums':[]
    },
    'output':0
})
```

```python
tests.append({
    'input':{
        'nums':[11]
    },
    'output':0
})
```

```python
tests.append({
    'input':{
        'nums':[1,2,3,4,5,6,7]
    },
    'output':0
})
```

```python
tests.append({
    'input':{
        'nums':[1,1,1]
    },
    'output':0
})
```

```python
def value(nums):
    position=0
    while position<len(nums):
        if position >0 and nums[position]<nums[position-1]:
            return position
        position+=1
    return 0
    
```

```python
for test in tests:
    print(value(**test['input'])==test['output'])
```

## optimize the algo 

Complexity of this is : O(N)

```python
## implement in binary search 
def valued(nums):
    low,high=0,len(nums)-1
    while low < high:
        mid = (low + high) // 2
        if nums[mid] > nums[high]:
            low = mid + 1
        else:
            high = mid-1

    return low if nums else 0
```

```python
for test in tests:
    print(valued(**test['input'])==test['output'])
```

```python

def search_rotation_dupli(nums,target):
    lo,hi=0,len(nums)-1

    while lo <= hi:
        mid=(lo+hi)//2

        if nums[mid]==target:
            return True

        if nums[lo]==nums[mid]==nums[hi]: # skip the duplicates
            lo+=1
            hi-=1
        elif nums[mid]<nums[hi]:
            if nums[lo] <=target<nums[mid]:
                hi=mid-1
            else:
                lo=mid+1
        else:
            if nums[mid]<target<=nums[hi]:
                lo=mid+1
            else:
                hi=mid-1
    return False

            
```
