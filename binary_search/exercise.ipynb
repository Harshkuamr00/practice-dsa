## implement the count_rotation function using the generic binary_search function 


```python
## given in question find the target in a array
#nums

def find(nums,target):
    lo,hi=0,len(nums)-1

    while lo <= hi:
        mid=(lo+hi)//2

        # checking conditions 
        if nums[mid] == target:
            return mid

        elif nums[mid] > target:
            hi=mid-1
        elif nums[mid] <target:
            lo=mid+1
    return -1
```

## rotationla problem 

### Given : a sorted array that has been rotated at an unknown pivot

we have to return index of the value after it rotated , otherwise return -1


```python
## following test case
tests=[]
tests.append({
    'input':{
        'nums':[4, 5, 6, 7, 0, 1, 2],
    },
    'output':4
})
```

```python
tests.append({
    'input':{
        'nums':[4, 5, 6, 7, 1, 2],
    },
    'output':4
})
```

```python
tests.append({
    'input':{
        'nums':[4, 5, 6, 7,-3,-2,-1, 0, 1, 2],
    },
    'output':4
})
```

```python
tests.append({
    'input':{
        'nums':[4, 5, 0, 1, 2],
    },
    'output':2
})
```

```python
tests.append({
    'input':{
        'nums':[],
    },
    'output':-1
})
```

```python
## code
def rotational(nums):
    lo,hi=0,len(nums)-1

    while lo<hi:
        mid=(lo+hi)//2

        if nums[mid]>nums[hi]:
            lo=mid+1
        else:
            hi=mid
    return lo if nums else -1
```

```python
for test in tests:
    print(rotational(**test['input'])==test['output'])
```
