## quick question 
given an array of integers nums sorted in ascending order find the starting and ending position of a given number 

### Define the test cases
following points contains test case:
1. if target matches mid
2. if list contains repeating values
3. if list does not contain value
4. if list first element contain the value
5. if list last element contain the value
6. if list is empty

```python
# test case 1
test ={
    'input':{
        'nums':[1,2,3,4,5,6,7],
        'target':4
    },
    'output':3
}
```

```python
tests=[]
tests.append(test)
tests.append({
    'input':{
        'nums':[1,2,3,4,5,6,7],
        'target':8
    },
    'output':-1
})
```

```python
tests.append({
    'input':{
        'nums':[1,1,2,2,3,3,4],
        'target':1
    },
    'output':0
})
```

```python
tests.append({
    'input':{
        'nums':[],
        'target':1
    },
    'output':-1
})
```

```python
tests.append(test)
test ={
    'input':{
        'nums':[1,1,2,2,3,3,4],
        'target':4
    },
    'output':6
}
```

```python
tests
```

```python
def test_locate(nums,target,mid):
    if nums[mid] == target:
        if mid > 0 and nums[mid-1] ==target:
            return 'left'
        return 'found'

    elif nums[mid] < target:
        return 'right'
    elif nums[mid] > target:
        return 'left'

def find(nums,target):
    low, high= 0, len(nums)-1
   

    while low <= high:
        mid = (low+high)//2
        result=test_locate(nums,target,mid)

        if result == 'mid':
            return mid

        elif result == 'right':
            low =mid +1

        elif result == 'left':
            high = mid -1
    return -1
    
        
            
        
```

```python
find(**test['input'])==test['output']
```

```python
for test in tests:
    print(find(**test['input'])==test['output'])
```
