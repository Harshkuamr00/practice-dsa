### pointer has type it depends where to start the lef and right all tpyes menstion in below:

1. Converging Pointers
-> start with ending of the list or array left end and right end, compare with specific condition if met then stop and stop until l1 and R1 cross each other.
used in pailaindrome

3. Parallel Pointers
-> in this both pointer start with same point like begning and move in the same direction right pointer is used to explore and find new information and left pointer is to track progress, maintain constraints.
used in substrings

5. Trigger-Based Pointers
-> in this apporach we move first pointer independently until it finds an element that meet a certain condition after this we start traversing with the secound pointer to addition information related to first pointer found.
this teqnice used when we have to process elements in stages

### Question 1: Two sum(sorted array)
Given a sorted array nums = [2, 7, 11, 15] and target = 9, return the indices of the two numbers that add up to target.

##### test Cases

```python
tests=[]
#test case 1
tests.append({
    "input":{
        "nums":[2,7,11,15],
        "target":9,
    },
    "output":[0,1]
})

# test case 2
tests.append({
    "input":{
        "nums":[2,3,4],
        "target":6,
    },
    "output":[0,2]
})
# test case 3
tests.append({
    "input":{
        "nums":[3,3],
        "target":6,
    },
    "output":[0,1]
})

# test case 4
tests.append({
    "input":{
        "nums":[3, 24, 50, 79, 88, 150, 345],
        "target":200,
    },
    "output":[2,5]
})

# test case 5
tests.append({
    "input":{
        "nums":[],
        "target":200,
    },
    "output":[]
})

# test case 6
tests.append({
    "input":{
        "nums":[3,3,3,3,3],
        "target":6,
    },
    "output":[0,1]
})

#test case 7
tests.append({
    "input":{
        "nums":[-1,0],
        "target":-1,
    },
    "output":[0,1]
})

    
```

```python
## code using conversing pointer
class solution:
    def __init__(self,nums,target):
        self.nums=nums
        self.target=target 
    ## two sum sorted array  
    def sol(self):
        if not self.nums:
            return []
        l,r=0,len(self.nums)-1

        ## taking loop
        while l<r:
            if self.nums[l]+self.nums[r]>self.target:
                r-=1
            elif self.nums[l]+self.nums[r]<self.target:
                l+=1
            else:
                result=[l,r]
                r-=1
                
        return result if result else []
        
                
                
                
```

```python
l1=solution([2,7,11,15], 9)
```

```python
print(l1.sol())
```

```python
for test in tests:
    current_test = solution(**test["input"])
    result = current_test.sol()
    print(f"Result: {result} | Expected: {test['output']} | Passed: {result == test['output']}")
```

### Question 2 : Remove Duplicates (in-place)
Brief: Modify the array in-place to remove duplicates and return the new length.

```python
tests=[]

tests.append({
    "input":{
        "nums":[1,1,2],
    },
    "Output":2
})

# test case 2
tests.append({
    "input":{
        "nums":[],
    },
    "Output":[]
})

# test case 3
tests.append({
    "input":{
        "nums":[-1,0,1,1],
    },
    "Output":3
})

# test case 4
tests.append({
    "input":{
        "nums":[1,2,3],
    },
    "Output":3
})
```

```python
## code 
class solution1:
    def __init__(self,nums):
        self.nums=nums

    # solution method
    def remove_dupilcates(self):
        l,r=0,len(self.nums)-1
        if not self.nums:
            return []

        while l<r:
            if self.nums[l]==self.nums[r]:
                self.nums.pop(l)
                l+=1
            elif self.nums[l]<self.nums[r]:
                r-=1
                
            else:
                r-=1
        return len(self.nums)      
            
```

```python
l=solution1([1,1,2])
```

```python
l.remove_dupilcates()
```

```python
tests
```

```python
for test in tests:
    current_test=solution1(**test['input'])
    result=current_test.remove_dupilcates()

    print(f"result: {result} | expected : {test['Output']} | Passed : {test['Output']==result}")
```
